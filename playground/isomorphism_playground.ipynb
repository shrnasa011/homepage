# Graph Isomorphism Playground
Explore color refinement and the Weisfeiler–Leman hierarchy with small graphs. Inspired by [GI Lab](https://github.com/shrnasa011/testing_isomorphism); this notebook is self-contained and does not access the private repository.

Run the cells from top to bottom.


```python
from collections import Counter
import networkx as nx
```

```python
def rook_graph(n=4):
    G=nx.Graph(); G.add_nodes_from((i,j) for i in range(n) for j in range(n))
    for u in G:
        for v in G:
            if u<v and (u[0]==v[0] or u[1]==v[1]): G.add_edge(u,v)
    return G

def shrikhande_graph():
    steps={(1,0),(3,0),(0,1),(0,3),(1,1),(3,3)}
    G=nx.Graph(); G.add_nodes_from((i,j) for i in range(4) for j in range(4))
    for i,j in G:
        for a,b in steps: G.add_edge((i,j),((i+a)%4,(j+b)%4))
    return G

def wl1_signature(G, rounds=8):
    colors={v:G.degree(v) for v in G}; trace=[]
    for _ in range(rounds):
        trace.append(tuple(sorted(Counter(colors.values()).items())))
        keys={v:(colors[v],tuple(sorted(colors[u] for u in G.neighbors(v)))) for v in G}
        palette={k:i for i,k in enumerate(sorted(set(keys.values()),key=repr))}
        colors={v:palette[keys[v]] for v in G}
    return tuple(trace+[tuple(sorted(Counter(colors.values()).items()))])

rook, shrikhande=rook_graph(),shrikhande_graph()
print('1-WL signatures match:',wl1_signature(rook)==wl1_signature(shrikhande))
print('Isomorphic:',nx.is_isomorphic(rook,shrikhande))
```

The rook and Shrikhande graphs are non-isomorphic but 1-WL collides on this pair. A mismatch proves non-isomorphism; matching signatures do not prove isomorphism.
